```python
import os

# Estructura de carpetas a crear dentro del Git Folder
folders = [
    "resources",
    "src/electrocasa/transformations",
    "src/electrocasa/utils",
    "notebooks"
]

# Crear directorios
for folder in folders:
    os.makedirs(folder, exist_ok=True)
    print(f"Directorio creado: {folder}")

# Archivos base vacíos para inicializar la estructura
files = [
    "databricks.yml",
    "resources/electrocasa_pipeline.yml",
    "resources/electrocasa_job.yml",
    "src/electrocasa/transformations/01_bronze.py",
    "src/electrocasa/transformations/02_silver.py",
    "src/electrocasa/transformations/03_gold.py",
    "src/electrocasa/utils/quality_checks.py",
    "notebooks/00_setup.ipynb",
    "notebooks/04_orchestrator_job.ipynb"
]

# Crear archivos vacíos si no existen
for file_path in files:
    if not os.path.exists(file_path):
        with open(file_path, "w", encoding="utf-8") as f:
            f.write("# Archivo del proyecto Electrocasa\n")
        print(f"Archivo creado: {file_path}")

print("\n¡Estructura de carpetas y archivos del Git Folder creada exitosamente!")
```
